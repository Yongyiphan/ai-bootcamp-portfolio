### Task 2.1 — The Base `chat()` Function

**Your Task:** Complete the `chat()` function. This single function will power **all** your bots — the only difference is the `system` prompt you pass in.

>  Key insight: same code + different system prompt = completely different bot.

```python
# Task 2.1 — Complete the base chat() function

def chat(prompt, system, temperature=0.7, max_tokens=200):
    """Ask the AI. Returns only the answer string."""
    response = client.chat.completions.create(
        model    = 'gpt-4o-mini',
        messages = [
            {'role': 'system', 'content': system},
            {'role': 'user', 'content': prompt}
        ],
        temperature = temperature,
        max_tokens  = max_tokens
    )
    return response.choices[0].message.content

# Quick test — should print something like: "2 + 2 equals 4."
print(chat('What is 2+2?', 'You are a helpful assistant.'))
```

### Task 3.1 — Prove the Memory Problem

**Your Task:** No blanks — just run the cell and observe what happens.

>  Replace `[YOUR NAME]` in the code with your actual name before running!

```python
# Task 3.1 — Prove the memory problem
# The first independent request tells the model Edgar's name.

# Call 1: tell the AI your name
r1 = client.chat.completions.create(
    model    = 'gpt-4o-mini',
    messages = [{'role': 'user', 'content': 'My name is Edgar Yong. Nice to meet you!'}],
    max_tokens = 50
)
print('Call 1 response:')
print(r1.choices[0].message.content)
print()

# Call 2: ask the AI your name — no history included!
r2 = client.chat.completions.create(
    model    = 'gpt-4o-mini',
    messages = [{'role': 'user', 'content': 'What is my name?'}],
    max_tokens = 50
)
print('Call 2 response (no history passed):')
print(r2.choices[0].message.content)
```

**What happened? Why? — double-click to edit:**

Did the AI remember your name in Call 2? **No. It was not given Edgar's name in the second request.**

Why does this happen? **OpenAI API calls are stateless. Each request only contains the messages explicitly sent with that request. Because Call 2 does not include Call 1's message, the model has no access to the name Edgar. Conversation memory must be created by sending previous messages again.**

### Task 3.2 — Build a Simple Memory Function

**Your Task:** Fill in the 3 blanks to make the function maintain conversation history.

The three steps, in order:
1. Add the user's question to `history` **before** the API call
2. Pass `history` as the `messages` argument
3. Add the AI's reply to `history` **after** the API call

> Replace `[YOUR NAME]` and `[YOUR MAJOR]` with your real details!

```python
# Task 3.2 — Build a memory function

history = [
    {'role': 'system', 'content': 'You are a friendly study assistant for NUS students.'}
]

def chat_with_memory(question):
    """Add question to history, call API, add answer to history, return answer."""

    # Step 1: Add user question to history BEFORE the API call
    history.append({'role': 'user', 'content': question})

    # Step 2: Call API with the FULL history
    response = client.chat.completions.create(
        model      = 'gpt-4o-mini',
        messages   = history,
        max_tokens = 150
    )
    answer = response.choices[0].message.content

    # Step 3: Add AI's reply to history AFTER getting the response
    history.append({'role': 'assistant', 'content': answer})

    total_tok = response.usage.total_tokens
    print(f'  [Turn {len(history)//2} | Messages in history: {len(history)} | Tokens this call: {total_tok}]')
    return answer


# Test the bot across four connected conversation turns.
turns = [
    'My name is Edgar and I am studying AI application development.',
    'What should I focus on in my degree?',
    'What was my name again?',
    'How can I use my degree in the Singapore tech industry?'
]

for q in turns:
    print(f'You: {q}')
    print(f'Bot: {chat_with_memory(q)}')
    print()
```

**Reflection — double-click to edit:**

Did the bot remember your name in Turn 3? **Yes. Edgar's name remains in the history passed to the model.**

What do you notice about the token count each turn? Why does it grow?

**The token count generally increases because every request includes the complete conversation history, including all earlier questions and answers.**

What problem might arise if a conversation went on for 1,000 turns?

**The requests could become slow and expensive or exceed the model's context limit. A production app should trim older messages or replace them with a summary.**

### Task 3.3 — The Full `MemoryBot` Class

**Your Task:** No blanks here — run the cell as-is and test it. Notice that `tutor` and `coach` have **completely separate** and independent histories.

```python
# Task 3.3 — MemoryBot class (no blanks — run and observe)

class MemoryBot:
    def __init__(self, personality='helpful'):
        prompts = {
            'helpful': 'You are a friendly, helpful assistant.',
            'tutor':   'You are a patient CS tutor. Use simple examples. Always check understanding.',
            'coach':   'You are an energetic life coach. Always motivate. Start every reply with an emoji.',
        }
        self.system    = prompts.get(personality, prompts['helpful'])
        self.history   = []
        self.turn      = 0
        self.total_tok = 0

    def ask(self, question):
        self.turn += 1
        self.history.append({'role': 'user', 'content': question})
        messages = [{'role': 'system', 'content': self.system}] + self.history
        response = client.chat.completions.create(
            model      = 'gpt-4o-mini',
            messages   = messages,
            max_tokens = 200
        )
        answer = response.choices[0].message.content
        self.total_tok += response.usage.total_tokens
        self.history.append({'role': 'assistant', 'content': answer})
        return answer

    def status(self):
        cost = self.total_tok * 0.0000002
        return f'Turn {self.turn} | {len(self.history)} messages | {self.total_tok} tokens | ${cost:.6f}'


# Test tutor personality
print('=== TUTOR BOT ===')
tutor = MemoryBot(personality='tutor')
print(tutor.ask('Explain what a variable is.'))
print(tutor.status())
print(tutor.ask('Can you give me a quiz question about variables?'))
print(tutor.status())
print()

# Test coach personality
print('=== COACH BOT ===')
coach = MemoryBot(personality='coach')
print(coach.ask('I feel like giving up on programming. It is too hard.'))
print(coach.status())
```
